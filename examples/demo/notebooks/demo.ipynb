# jl-forge demo

Runs in both the xeus-python and the Pyodide site. `jl_forge_demo` is mounted from source in xeus and installed from its wheel in Pyodide.

The last cell checks the ipywidgets round trip: clicking the button updates a label and two `Output` widgets from the kernel.

```python
from jl_forge_demo import kernel_info

kernel_info()
```

```python
import ipywidgets as widgets

button = widgets.Button(description="Click me")
label = widgets.Label("not clicked")
captured = widgets.Output()  # `with captured:` from a callback; empty in xeus-python, voici#127
appended = widgets.Output()  # append_stdout() sets Output.outputs directly; works in both kernels


def on_click(_):
    label.value = "clicked"
    with captured:
        print("captured from the callback")
    appended.append_stdout("appended from the callback\n")


button.on_click(on_click)
widgets.VBox([button, label, captured, appended])
```
